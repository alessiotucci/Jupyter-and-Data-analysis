## 6. Grouping and aggregation
Grouping and aggregation can be very useful to gain insight, we will take a look at how the price and reviews score vary by room_type

````python
df.groupby("room_type")
````

Using only the `groupby()` function will return just an object, we need to do more...

### 6.1 Grouping by room_type, and calculate the avg_rating

````python
# 1) Group by room_type
# 2) Compute the average review score for each group
avg_rating = df.groupby("room_type")["review_scores_rating"].mean()
avg_rating
````

### 6.2 Grouping by room_type, and calculcate the avg_price

````python
#1) Groub by room type
#2) Compute the average price for each group
avg_price = df.groupby("room_type")["price"].mean()
avg_price
````

### 6.3 Grouping and counting unique values
This was in the practice test, so it might be useful to learn some practical way to count unique values

````python
# 1) How many unique host are in the dataset
n_hosts = df["host_id"].nunique()
n_hosts
````

````python
# 2) How many listings does each host have (one row per host)
listing_per_host = df.groupby("host_id").size()
listing_per_host
````

````python
# 3) sort the host that have the most listing
top_hosts = listing_per_host.sort_values(ascending=False)
top_hosts.head() # top 5
````

## 8. Categorical variables
If we run `df.dtypes` we can see if the text columns are treated like object and if so, we can try to convert them to **category**

````python
df.dtypes
````

````python
df["room_type"] = df["room_type"].astype("category") # basic conversion
df.dtypes # check after we update the dataframe
````

### 8.1 Categorical ordinal sequences
Now let's try to convert the maximum stay in a category that hava a natural sequences. 
To do this, we will create a natural sequences in the **maximum stay category**

````python
# 1) first, create a var with the wanted order
order = ["Less than a month", "1 to 2 months", "2 to 12 months", "1 to 3 years", "3 years +"]

df["maximum_stay_cat"] = pd.Categorical(df["maximum_stay"], categories=order, ordered=True)
# with this we can check the difference
df[["maximum_stay", "maximum_stay_cat"]].dtypes
````
