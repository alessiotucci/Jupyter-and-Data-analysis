Let's take now a look at a sample question of the *Practice Quiz*.

```How many variables contain missing values?```

To answer this question we can use the 

````python
df.isnull() # this return a new dataframe with true = 1 and  false = 0, but it's not very useful now...
````

````python
df.isnull().sum() # This will sum the value of each column, for example for id we get 0 since there is not missing value
````

````python
missing_counts = df.isnull().sum() 
missing_counts[missing_counts > 0] # we can use the square brackets to filter only when it's true
````

````python
len(missing_counts[missing_counts > 0]) # the final answer
````
